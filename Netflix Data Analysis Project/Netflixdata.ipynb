```python
# Dirty Values - N/A, NULL, Blank, spaces

df.replace(                                                 # Data me ["N/A", "NULL", "", " "] ye values jaha bhi hongi unhe "NAN" matlab "Blank" se replace kar dega
    ["N/A", "NULL", "", " "], np.nan, inplace=True
)
```

```python
# Extra spaces remove karenge
                                                                                                                            # apply ek function he jisse apply ke andar likha hua function pure data par apply ho jayega
df = df.apply(lambda x: x.str.strip().str.title() if x.dtype == "string" else x)                                            # lambda use hota he koi function banane ke liye ( lambda ke aage likhi hui line ek function ki tarah kaam karegi)
                                                                                                                            # isme x wo data he jo cell me likha hua hoga, x.str.strip() data se unwanted space hateyega or x.dtype == "object" check karega ki sirf text data se hi sapce hate agar x ka datatype text nahi he to koi change nahi karna he
                                                                                                                            # str.title() data ko Standardize karta he(matlab first alphabet ko capital or baki small)
```

```python
# Numeric Columns

numeric_col = ["Age", "Watch_Hours", "Monthly_Fee", "Movies_Watched", "Episodes_Watched", "Avg_Rating", "Monthly_Sessions", "Download_Count"]
```

```python
for col in numeric_col:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df
```

```python
df.describe()
```

```python
df=df[(df["Age"]>=18) & (df["Age"]<=100)]
```

```python
df.describe()           # ab age data clean hua he (ab min age ki value negative nahi he)
```

```python
# Watch hours data ko bhi positive karenge

df = df[df["Watch_Hours"]>=0]
```

```python
df.describe()
```

```python
# Monthly fee ko bhi postitve karenge

df = df[df["Monthly_Fee"]>=0]
```

```python
df.describe()
```

```python
# Date column me se time hatana padega or null cells ko fill karna hoga nan se taki koi error na aaye

df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
df["Last_Watch_Date"] = pd.to_datetime(df["Last_Watch_Date"], errors = "coerce")
```

```python
df
```

```python
df[["Join_Date", "Last_Watch_Date"]].isnull().sum()
```

```python
# Missing Values Handle Karna

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Watch_Hours"] =  df["Watch_Hours"].fillna(df["Watch_Hours"].mean())
df["Monthly_Fee"] = df["Monthly_Fee"].fillna(df["Monthly_Fee"].mean())
```

```python
# Categorical Column Handle karenge

df["Genre"] = df["Genre"].fillna("Unknown")
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
df["Device"] = df["Device"].fillna("Unknown")
```

```python
df.shape
```

```python
# Feature Engineering(Calculated Column add karna)

# Customer Revenue Column add karenge

df["Customer_Revenue"] = (df["Monthly_Fee"] * df["Monthly_Sessions"])
```

```python
df
```

```python
# Engagement Score (ab ye column add karenge)

df["Engagement Score"] = (df["Watch_Hours"]+df["Monthly_Sessions"]+df["Download_Count"])
```

```python
df.head()
```

```python
# Watch Category

bins = [-1,20,50,100,np.inf]                         # bins matlab 0-20,20-50,50-100 or 100 -infinite watch hours ke liye Watch_Category me kya labels dene hain 
label = ["Low","Medium","High","Very High"]

df["Watch_Category"] = pd.cut(df["Watch_Hours"], bins=bins, labels=label)   # Watch_Category column add ho jayega
```

```python
df.head()
```

```python
# Customer Age Group

bins = [17,25,35,50,100]                                # Age column ke hisab se bins banayenge
label = ["18-25", "26-35", "36-50", "51-100"]

df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=label)        # Age_Group column ban jayega
```

```python
df.head()
```

```python
df
```

```python
df.shape
```

```python
# Account Active Flag                   Account_Status Column he pehle se fir bhi humne Active_Flag column banaya he

df["Active_Flag"] = np.where(df["Account_Status"]=="Active", 1, 0)
```
